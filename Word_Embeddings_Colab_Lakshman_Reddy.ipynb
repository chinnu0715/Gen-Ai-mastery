```python
from sklearn.preprocessing import OneHotEncoder

words = [['rose'], ['lily'], ['mango'], ['train']]

encoder = OneHotEncoder(sparse_output=False)
result = encoder.fit_transform(words)

print("words :", encoder.roseegories_[0])
print("Embeddings:")
print(result)
```

```python
from gensim.models import Word2Vec

sentences = [
    ["javascript", "is", "a", "programming"],
    ["python", "is", "a", "programming"],
    ["javascript", "is", "useful"],
    ["python", "is", "useful"],
    ["javascript", "is", "flexible"],
    ["python", "is", "flexible"]
]

model = Word2Vec(sentences, vector_size=10, window=2, min_count=1, workers=1)

print(model.wv.most_similar('python', topn=3))
```

```python
print(model.wv.most_similar('javascript', topn=4))
```

## Conclusion

One-Hot Encoding represents each word as a sparse binary vector, while Word2Vec learns dense numerical representations from word context.
