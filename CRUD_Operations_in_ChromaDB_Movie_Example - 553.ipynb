```python
import chromadb

# Create ChromaDB client
client = chromadb.Client()

# Create collection
collection = client.create_collection(name="movie_demo")

print("Collection created:", collection.name)
print("Current document count:", collection.count())
```

```python
collection.add(
    documents=[
        "The Lion King is an animated movie about a young lion becoming king.",
        "Finding Nemo follows a clownfish searching for his missing son.",
        "Interstellar explores space travel, time, and the survival of humanity.",
        "The Avengers brings superheroes together to protect the world.",
        "Toy Story follows toys that come to life when people are not around."
    ],

    metadatas=[
        {"category": "animation"},
        {"category": "animation"},
        {"category": "science_fiction"},
        {"category": "superhero"},
        {"category": "animation"}
    ],

    ids=[
        "movie1",
        "movie2",
        "movie3",
        "movie4",
        "movie5"
    ]
)

print("Documents added!")
print("Total count:", collection.count())
```

```python
preview = collection.peek(limit=3)

for doc_id, doc in zip(
    preview["ids"],
    preview["documents"]
):
    print(doc_id, "->", doc)
```

```python
result = collection.get(
    ids=["movie1", "movie3"]
)

for doc_id, doc, meta in zip(
    result["ids"],
    result["documents"],
    result["metadatas"]
):
    print(f"{doc_id} | {meta} | {doc}")
```

```python
results = collection.query(
    query_texts=[
        "stories about space and the future of humanity"
    ],
    n_results=2
)

for doc_id, doc, distance in zip(
    results["ids"][0],
    results["documents"][0],
    results["distances"][0]
):
    print(
        f"{doc_id} | distance={distance:.4f} | {doc}"
    )
```

```python
results = collection.query(
    query_texts=[
        "fun animated stories for families"
    ],
    n_results=3,
    where={"category": "animation"}
)

for doc_id, doc, distance in zip(
    results["ids"][0],
    results["documents"][0],
    results["distances"][0]
):
    print(
        f"{doc_id} | distance={distance:.4f} | {doc}"
    )
```

```python
collection.update(
    ids=["movie3"],

    documents=[
        "Interstellar explores space exploration, black holes, time, and humanity's future."
    ],

    metadatas=[
        {
            "category": "science_fiction",
            "updated": True
        }
    ]
)

updated = collection.get(ids=["movie3"])

print(updated["documents"])
print(updated["metadatas"])
```

```python
collection.upsert(
    documents=[
        "The Avengers features a team of superheroes defending Earth.",
        "The Incredibles follows a family of superheroes balancing family life and crime-fighting."
    ],

    ids=[
        "movie4",
        "movie6"
    ]
)

print("Total count after upsert:", collection.count())

print(
    collection.get(
        ids=["movie4", "movie6"]
    )["documents"]
)
```

```python
collection.delete(
    ids=["movie2"]
)

print(
    "Total count after delete:",
    collection.count()
)

print(
    "Remaining ids:",
    collection.get()["ids"]
)
```

```python
collection.delete(
    where={"category": "animation"}
)

print(
    "Total count after category delete:",
    collection.count()
)

print(
    "Remaining ids:",
    collection.get()["ids"]
)
```

```python
final_state = collection.get()

print(
    "Final document count:",
    collection.count()
)

for doc_id, doc, meta in zip(
    final_state["ids"],
    final_state["documents"],
    final_state["metadatas"]
):
    print(f"{doc_id} | {meta} | {doc}")
```
